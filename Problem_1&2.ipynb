```python
'''Student Grade Manager'''
def calculate_stats(scores):
    total=sum(scores)
    average=total/len(scores)
    highest=max(scores)
    lowest=min(scores)
    return(total,average,highest,lowest)

def assign_grade(average):
    if average>=80:
        return "A"
    elif average>=65:
        return "B"
    elif average>=50:
        return "C"
    elif average>=35:
        return "D"
    else:
        return "F"

student_name=input("Enter Name:")
Num_of_sub=int(input())
print(f"Welcome {student_name}")
print(f"Enter number of subjects:")
sub_list=[]
for i in range(Num_of_sub):
    num=float(input())
    if num>=0 and num<=100:
        sub_list.append(num)
    else:
        print("Number is invalid and give valid number")
        while num<0 or num>100:
            num=float(input())
        sub_list.append(num)
pass_sub=[sub_num for sub_num in sub_list if sub_num>=50]
total,avg,high,low=calculate_stats(sub_list)
print(f"Student Name:{student_name}\nScores:{sub_list}\nTotal:{total}\nAverage:{avg:.2f}\nHighest:{high} | Lowest:{low}\nGrade:{assign_grade(avg)}\nPassed:{len(pass_sub)} out of {Num_of_sub} subjects")

```
